### BLOQUE 1: ASIGNAR ACCIDENTES A  CLUSTERS/POLIGONOS DE VORONOI 

```python
# ===============================
# 1. Crear GeoDataFrame de accidentes
# ===============================
df_acc = df_acc.dropna(subset=["Latitud", "Longitud"])

gdf_acc = gpd.GeoDataFrame(
    df_acc,
    geometry=gpd.points_from_xy(df_acc["Longitud"], df_acc["Latitud"]),
    crs="EPSG:4326"
)

# ===============================
# 2. KMeans clustering
# ===============================
N_CLUSTERS = 100
X = np.vstack([gdf_acc.geometry.x.values, gdf_acc.geometry.y.values]).T
kmeans = KMeans(n_clusters=N_CLUSTERS, random_state=42, n_init=10)
labels = kmeans.fit_predict(X)

gdf_acc["cluster_id"] = labels
centers_xy = kmeans.cluster_centers_

# ===============================
# 3. Voronoi polígonos finitos
# ===============================
def voronoi_finite_polygons_2d(vor, radius=None):
    new_regions, new_vertices = [], []
    center = vor.points.mean(axis=0)
    if radius is None:
        radius = np.ptp(vor.points, axis=0).max() * 2  

    all_ridges = {}
    for (p1, p2), (v1, v2) in zip(vor.ridge_points, vor.ridge_vertices):
        all_ridges.setdefault(p1, []).append((p2, v1, v2))
        all_ridges.setdefault(p2, []).append((p1, v1, v2))

    for p1, region in enumerate(vor.point_region):
        vertices = vor.regions[region]
        if all(v >= 0 for v in vertices):
            new_regions.append(vertices)
            continue

        ridges = all_ridges[p1]
        new_region = [v for v in vertices if v >= 0]

        for p2, v1, v2 in ridges:
            if v2 < 0: v1, v2 = v2, v1
            if v1 >= 0: continue

            t = vor.points[p2] - vor.points[p1]
            t /= np.linalg.norm(t)
            n = np.array([-t[1], t[0]])
            midpoint = vor.points[[p1, p2]].mean(axis=0)
            direction = np.sign(np.dot(midpoint - center, n)) * n
            far_point = vor.vertices[v2] + direction * radius

            new_region.append(len(new_vertices))
            new_vertices.append(far_point)

        vs = np.asarray([vor.vertices[v] if v >= 0 else new_vertices[v] for v in new_region])
        c = vs.mean(axis=0)
        angles = np.arctan2(vs[:,1] - c[1], vs[:,0] - c[0])
        new_region = [v for _, v in sorted(zip(angles, new_region))]
        new_regions.append(new_region)

    return new_regions, np.vstack([vor.vertices, new_vertices])

# aplicar Voronoi
vor = Voronoi(centers_xy)
regions, vertices = voronoi_finite_polygons_2d(vor)

# construir polígonos
polygons = []
for region in regions:
    polygon = vertices[region]
    polygons.append(Polygon(polygon))

gdf_clusters = gpd.GeoDataFrame(
    {"cluster_id": range(len(polygons))},
    geometry=polygons,
    crs="EPSG:4326"
)

# ===============================
# 4. Recortar Voronoi al límite real de Barcelona (OSM)
# ===============================
gdf_barcelona = ox.geocode_to_gdf("Barcelona, Spain").to_crs("EPSG:4326")
gdf_clusters_clipped = gpd.overlay(gdf_clusters, gdf_barcelona, how="intersection")

print("Clusters después del recorte con Barcelona:", len(gdf_clusters_clipped))

# ===============================
# 5. Visualizar con Folium
# ===============================
m = folium.Map(location=[41.3879, 2.16992], zoom_start=12, tiles="cartodbpositron")

# añadir polígonos de clusters
for _, row in gdf_clusters_clipped.iterrows():
    if row.geometry.is_empty: 
        continue
    sim_geo = gpd.GeoSeries(row.geometry).simplify(0.0005)
    geo_j = sim_geo.to_json()
    folium.GeoJson(
        data=geo_j,
        style_function=lambda x: {
            "fillColor": "blue",
            "color": "black",
            "weight": 0.5,
            "fillOpacity": 0.2
        }
    ).add_to(m)

# añadir accidentes de muestra
for _, row in gdf_acc.sample(10000).iterrows():
    folium.CircleMarker(
        location=[row["Latitud"], row["Longitud"]],
        radius=2,
        color="red",
        fill=True,
        fill_opacity=0.6
    ).add_to(m)

# guardar mapa en html
map_filename = "clusters_barcelona_real.html"
m.save(map_filename)
print(f"✅ Mapa guardado como {map_filename}")
```

### BLOQUE 2: Construir candidatos en la red viaria (para no-accidentes)

```python
# ===============================
# BLOQUE 6. Construir candidatos en la red viaria (para no-accidentes)
# ===============================

# Descargar red viaria de Barcelona
G = ox.graph_from_polygon(gdf_barcelona.geometry.iloc[0], network_type="drive")
nodes, edges = ox.graph_to_gdfs(G)
edges = edges.to_crs("EPSG:25831")  # UTM metros para discretización

# Asignar cluster a cada tramo (intersecta cada tramo de carretera con los polígonos de Voronoi, para asignar un cluster_id a cada tramo)
gdf_clusters_m = gdf_clusters_clipped.to_crs("EPSG:25831")
edges_clusters = gpd.overlay(edges, gdf_clusters_m[["cluster_id","geometry"]], how="intersection")

# Función para generar puntos cada N metros (incluyendo extremos -> intersecciones). Esto permite crear candidatos a “no-accidentes” a lo largo de la red.
def points_along_line(line, step=40):
    L = line.length
    distances = np.arange(0, L+1e-6, step)
    return [line.interpolate(d) for d in distances]

# Generar candidatos
records = []
for idx, row in edges_clusters.iterrows():
    pts = points_along_line(row.geometry, step=40)  # cada 40 m
    for p in pts:
        records.append({
            "cluster_id": row["cluster_id"],
            "geometry": p,
            "tramo": row["osmid"]  # aquí guardamos el id del tramo
        })

gdf_candidates = gpd.GeoDataFrame(records, crs=edges_clusters.crs).to_crs("EPSG:4326")
print("Candidatos sobre la red viaria:", len(gdf_candidates))

# ===============================
# 7. Generar no-accidentes por hora y cluster
# ===============================
gdf_acc = gdf_acc.to_crs("EPSG:4326")  # aseguramos mismo CRS
gdf_acc["time_n"] = pd.to_datetime(gdf_acc["time_n"], errors="coerce")


BUF_M = 15   # distancia mínima a accidente real
N_NEG = 2    # nº de negativos por (hora,cluster)

# Proyección a metros para buffer
gdf_acc_m = gdf_acc.to_crs("EPSG:25831")
gdf_cand_m = gdf_candidates.to_crs("EPSG:25831")

neg_records = []

# Iterar sobre cada (hora, cluster_id) con accidentes
# Aseguramos que ningún no-accidente generado esté a menos de 15 m de un accidente que ocurrió en la misma fecha y cluster
for (hour, cid), acc_grp in gdf_acc.groupby(["time_n","cluster_id"]):
    acc_buf = acc_grp.to_crs("EPSG:25831").buffer(BUF_M).unary_union
    cand = gdf_cand_m[gdf_cand_m["cluster_id"] == cid]
    if len(cand) == 0:
        continue
    mask_far = ~cand.geometry.intersects(acc_buf)
    cand_ok = cand[mask_far]
    if len(cand_ok) == 0:
        continue
    sampled = cand_ok.sample(min(N_NEG, len(cand_ok)))
    sampled = sampled.to_crs("EPSG:4326")
    sampled["time_n"] = hour
    sampled["accident"] = 0
    neg_records.append(sampled)

if neg_records:
    gdf_neg = pd.concat(neg_records, ignore_index=True)
else:
    gdf_neg = gpd.GeoDataFrame(columns=["geometry","time_n","accident","cluster_id","tramo"], crs="EPSG:4326")

# ===============================
# 8. Unir accidentes y no-accidentes
# ===============================
if "tramo" not in gdf_acc.columns:
    gdf_acc["tramo"] = np.nan  # si no existiera, añadimos columna

dataset = pd.concat([
    gdf_acc[["time_n","accident","geometry","cluster_id","tramo"]],
    gdf_neg[["time_n","accident","geometry","cluster_id","tramo"]]
], ignore_index=True)

print("✅ Dataset final con tramo incluido:", dataset.shape)
```

### BLOQUE 4. RELLENAR LAS HORAS QUE NOS FALTAN CON UN 1 ACCIDENTE POR CLUSTER/HORA

```python
### BLOQUE 7
# ===============================
# 1. Definir rango completo de horas
# ===============================
start = pd.Timestamp("2017-01-01 00:00:00")
end   = pd.Timestamp("2024-12-31 23:00:00")
full_hours = pd.date_range(start=start, end=end, freq="H")

# ===============================
# 2. Horas ya presentes (con accidente o no accidente generado por buffer)
# ===============================
existing_hours = pd.to_datetime(dataset["time_n"]).dt.floor("h").unique()

# ===============================
# 3. Horas faltantes = horas sin ningún accidente
# ===============================
missing_hours = np.setdiff1d(full_hours, existing_hours)
missing_hours = pd.to_datetime(missing_hours)

print("⏳ Horas totales esperadas:", len(full_hours))
print("✅ Horas ya cubiertas (con accidente/no accidente):", len(existing_hours))
print("❌ Horas sin accidentes:", len(missing_hours))

# ===============================
# 4. Clusters disponibles
# ===============================
clusters = dataset["cluster_id"].unique()
print("Nº de clusters:", len(clusters))

# ===============================
# 5. Generar un no-accidente por cluster en cada hora faltante
# ===============================
# escoger 1 candidato aleatorio por cluster
cluster_to_candidate = (
    gdf_candidates.groupby("cluster_id")
    .apply(lambda x: x.sample(1, random_state=42).iloc[0])   # reproducible
)

# expandir horas faltantes × clusters
idx = pd.MultiIndex.from_product([missing_hours, clusters],
                                 names=["time_n", "cluster_id"])
df_missing = idx.to_frame(index=False)

# unir con la geometría fija de cada cluster
df_missing = df_missing.merge(
    cluster_to_candidate[["geometry"]],
    left_on="cluster_id", right_index=True, how="left"
)

df_missing["accident"] = 0
df_missing = gpd.GeoDataFrame(df_missing, geometry="geometry", crs=gdf_candidates.crs)

print("No-accidentes añadidos (horas sin accidentes):", len(df_missing))

# ===============================
# 6. Concatenar con dataset original
# ===============================
dataset_final = pd.concat([dataset, df_missing], ignore_index=True)

print("📊 Dataset final:", dataset_final.shape)
```
